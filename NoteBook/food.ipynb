```python
from pyspark.sql import SparkSession
from pyspark.sql import types as T
from pyspark.sql import functions as F
import os
import sys
import glob
os.environ.pop("SPARK_HOME", None)
os.environ.pop("CLASSPATH", None)
```

```python
import sys
import glob

PROJECT_ROOT = "/home/yogavarman/Projects/FoodChain"

sys.path.insert(0, PROJECT_ROOT)

jars = glob.glob("/opt/spark/jars/*.jar")

spark = (
    SparkSession.builder
    .appName("Clean_users")
    .config("spark.jars", ",".join(jars))
    .getOrCreate()
)
```

```python
import pandas as pd
from pathlib import Path

folderpath = Path("/home/yogavarman/Projects/FoodChain/DataSet/RawData")
filename = "food"

xlsx_path = folderpath / f"{filename}.xlsx"
csv_path = folderpath / f"{filename}.csv"

# Read the specific sheet
df_excel = pd.read_excel(xlsx_path, sheet_name="food")  # or sheet_name=0


# Write to CSV
df_excel.to_csv(csv_path, index=False)

print("CSV created at:", csv_path)
```

```python
df =  (
    spark.read
    .format("csv")
    .option("header",True)
    .option("inferSchema",False)
    .load(str(csv_path))
)
```

```python

df=df.withColumn("food_name",F.trim(F.col("item")))\
        .withColumn("food_type",F.trim(F.col("veg_or_non_veg")))

df=df.drop("f_id","item","veg_or_non_veg")
df.printSchema()
```
